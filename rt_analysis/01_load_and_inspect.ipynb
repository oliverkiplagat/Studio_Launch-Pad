## 1. Setup & Load

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 150)

INFO_PATH = 'rt_movie_info_tsv.gz'

info = pd.read_csv(INFO_PATH, sep='\t', compression='gzip', encoding='latin-1')

print(f"rt_movie_info: {info.shape[0]:,} rows x {info.shape[1]} cols")
```
