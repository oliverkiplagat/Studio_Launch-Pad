# 02 Check Duplicate Rows


## 1. Load raw data

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 150)

INFO_PATH = 'rt_movie_info_tsv.gz'

info = pd.read_csv(INFO_PATH, sep='\t', compression='gzip', encoding='latin-1')

print(f"rt_movie_info: {info.shape[0]:,} rows")
```

## 2. Duplicate checks

```python
print("Full-row duplicates in rt_movie_info:", info.duplicated().sum())
print("Duplicate movie ids in rt_movie_info:  ", info['id'].duplicated().sum())
```

```python
# Show any duplicate rows found (if none, this will just be an empty table)
info[info.duplicated(keep=False)]
```

**Findings:**
- `rt_movie_info` has **zero** duplicate rows and zero duplicate `id`s — nothing to drop.


## 3. Export deduplicated data for the next notebook

```python
info_deduped = info.copy()  # no duplicates found, kept for consistent naming downstream
info_deduped.to_csv('rt_movie_info_deduped.csv', index=False)

print(f"Exported rt_movie_info_deduped.csv ({info_deduped.shape[0]:,} rows)")
print("Next: 03_clean_runtime_genre.ipynb")
```
